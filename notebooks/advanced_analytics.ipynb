# Advanced Business Insights

## Insight 1 – Highest Downside Risk (VaR Analysis)

Historical Value at Risk (95%) analysis shows that a small number of mutual funds exhibit significantly larger potential daily losses than the rest of the portfolio universe. These funds require higher risk tolerance and are better suited for aggressive investors. The VaR and CVaR rankings help identify schemes exposed to extreme downside risk.

**Supporting Chart:** `top10_var95.png`

---

## Insight 2 – Investor Cohort Behaviour

Investor cohort analysis reveals that recent investor cohorts contribute a larger share of total investments than earlier cohorts. This indicates strong growth in mutual fund adoption and increasing retail participation. The most preferred schemes vary across cohorts, reflecting changing investor preferences over time.

**Supporting Chart:** `investor_cohort_total_investment.png`

---

## Insight 3 – SIP Continuity Analysis

Investors with an average gap greater than 35 days between SIP transactions were classified as **At-Risk**. Monitoring these investors can help fund houses improve retention through reminders, engagement campaigns, or advisory support before SIP discontinuation occurs.

**Supporting Charts:**
- `sip_continuity_status.png`
- `sip_gap_distribution.png`

## Insight 4 – Portfolio Concentration (HHI)

The Herfindahl–Hirschman Index (HHI) highlights differences in portfolio diversification across equity funds. Funds with higher HHI values have larger exposure to a limited number of sectors or holdings, while lower HHI values indicate better diversification and reduced concentration risk.

**Supporting Chart:** `sector_hhi_concentration.png`

---

## Insight 5 – Risk-Adjusted Fund Performance

Sharpe Ratio ranking identifies funds delivering the highest return per unit of risk. Combining Sharpe Ratio with Alpha, Maximum Drawdown, and CAGR provides a comprehensive performance scorecard that helps investors compare funds beyond simple return percentages.

## Overall Conclusion

The analytics pipeline integrates NAV history, portfolio holdings, investor transactions, benchmark indices, and SIP data to provide a comprehensive evaluation of mutual fund performance. By combining risk metrics (VaR, CVaR, Sharpe Ratio, HHI), investor behaviour (cohort analysis and SIP continuity), and performance analytics (CAGR, Alpha, Beta, Benchmark Comparison), the platform enables data-driven fund selection and investment decision-making.
